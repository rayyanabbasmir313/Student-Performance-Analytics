# 02 - Data Cleaning 
- This notebook prepares the raw data for analysis: checking for duplicates, checking value ranges, and handling the G3 = 0 students.

```python
import pandas as pd

raw = pd.read_csv("../data/raw/student-por.csv", sep=";")
df = raw.copy()
```

```python
df.duplicated().sum()
```

### Duplicate Check

No duplicate rows were found in the dataset.

```python
df[['age', 'studytime', 'absences', 'Medu', 'Fedu', 'G1', 'G2', 'G3']].describe()
```

```python
df['school'].value_counts()
```

### Range Checks

- `G1` has a minimum of 0. One student (row 0, GP) has G1 = 0 but G2 = 11 and G3 = 11, so this 0 is likely a recording error. It is kept because it does not affect G3, the main outcome.

### Handling G3 = 0

There are 15 students with a final grade (G3) of 0. These students all have 0 absences, while some have non-zero G1 and G2 grades, suggesting that the zero may not represent their actual academic performance.

For this analysis, these students will be removed so that the cleaned dataset focuses on students with an observed final grade.

```python
df.shape
```

```python
df = df[df['G3'] != 0]
```

```python
df.shape
```

```python
df['G3'].min()
```

```python
(df['G2'] == 0).sum()
```

```python
df = df.reset_index(drop=True)
```

```python
import os

os.makedirs('../data/processed', exist_ok=True)

df.to_csv('../data/processed/student-por-clean.csv', index=False)
```

## Cleaning Summary

Cleaning complete: 649 → 634 rows. No duplicates were found. 15 students with G3 = 0 were removed. The cleaned dataset was saved to `data/processed/student-por-clean.csv`.
