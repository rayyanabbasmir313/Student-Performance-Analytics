## Data Understanding Notes

- Missing values: none. All 33 columns contain 649 non-null values.
- Data types: 16 numeric columns and 17 text columns.
- `studytime` is an ordinal code, not a number of hours:
  - 1 = less than 2 hours per week
  - 2 = 2 to 5 hours per week
  - 3 = 5 to 10 hours per week
  - 4 = more than 10 hours per week
- `Medu` and `Fedu` are ordinal education-level codes:
  - 0 = none
  - 1 = primary education
  - 2 = 5th to 9th grade
  - 3 = secondary education
  - 4 = higher education
- failures: 0 = never failed, 1 = once, 2 = twice, 3 = three or more
  - The documentation says 4, but the data's highest value is 3
  - 85% of students have 0 failures, so the groups are very unbalanced
### G3 (final grade):
- 15 students have G3 = 0 (out of 649)
- They all had real G1 grades, so the 0 likely means they didn't finish, not that they scored zero
- Their absences are all 0, which doesn't match someone who left, so absences may not have been recorded
- 14 of the 15 are from MS
- This could affect research questions 2 (absences) and 5 (schools)
